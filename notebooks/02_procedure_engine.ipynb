```python
class ProcedureEngine:

    STOPWORDS = {
        "what", "when", "where", "why", "how",
        "is", "are", "was", "were", "a", "an", "the",
        "if", "of", "in", "on", "at", "to", "for",
        "and", "or", "but", "can", "could", "would",
        "should", "do", "does", "did", "my", "me",
        "i", "we", "they", "student", "students",
        "about", "after", "before", "get", "know",
        "many", "much", "must"
    }

    def __init__(self, procedures, deadlines=None):
        self.procedures = procedures
        self.deadlines = deadlines or []

    @staticmethod
    def normalize(text):
        return re.sub(
            r"[^a-z0-9\s]",
            " ",
            str(text).lower()
        ).strip()

    def tokenize(self, text):
        return {
            word
            for word in self.normalize(text).split()
            if word not in self.STOPWORDS and len(word) > 1
        }

    def _search_records(
        self,
        query,
        records,
        id_field,
        name_field,
        keyword_field,
        top_k=5
    ):

        query_normalized = self.normalize(query)
        query_terms = self.tokenize(query)

        if not query_terms:
            return []

        results = []

        for record in records:

            name = record.get(name_field, "")
            description = record.get("description", "")
            keywords = record.get(keyword_field, [])

            name_terms = self.tokenize(name)
            description_terms = self.tokenize(description)

            matched_terms = set()
            score = 0

            # Match procedure name
            name_overlap = query_terms.intersection(name_terms)

            score += len(name_overlap) * 4.0
            matched_terms.update(name_overlap)

            # Match description
            description_overlap = query_terms.intersection(
                self.tokenize(description)
            )

            score += len(description_overlap) * 0.5
            matched_terms.update(description_overlap)

            # Match retrieval keywords
            for keyword in keywords:

                keyword_normalized = self.normalize(keyword)
                keyword_terms = self.tokenize(keyword)

                overlap = query_terms.intersection(keyword_terms)

                if overlap:
                    score += len(overlap) * 3.0
                    matched_terms.update(overlap)

                    if keyword_normalized in query_normalized:
                        score += 5.0

            coverage = len(matched_terms) / len(query_terms)
            final_score = score * coverage

            if final_score > 0:

                results.append({
                    **record,
                    "retrieval_score": round(final_score, 4),
                    "query_coverage": round(coverage, 4),
                    "matched_terms": sorted(matched_terms)
                })

        results.sort(
            key=lambda item: (
                item["retrieval_score"],
                item["query_coverage"]
            ),
            reverse=True
        )

        return results[:max(0, top_k)]

    # Search descriptive procedures
    def search_procedures(self, query, top_k=5):

        return self._search_records(
            query=query,
            records=self.procedures,
            id_field="procedure_id",
            name_field="procedure_name",
            keyword_field="procedure_keywords",
            top_k=top_k
        )

    # Search structured deadlines
    def search_deadlines(self, query, top_k=3):

        return self._search_records(
            query=query,
            records=self.deadlines,
            id_field="deadline_id",
            name_field="action",
            keyword_field="keywords",
            top_k=top_k
        )

    # Search both procedures and deadlines
    def search(self, query, top_k=5):

        procedure_results = self.search_procedures(
            query, top_k=top_k
        )

        deadline_results = self.search_deadlines(
            query, top_k=top_k
        )

        return {
            "procedures": procedure_results,
            "deadlines": deadline_results
        }

    # Retrieve a procedure using its ID
    def get_procedure_by_id(self, procedure_id):

        for procedure in self.procedures:

            if procedure["procedure_id"] == procedure_id:
                return procedure

        return None

    # Retrieve a deadline using its ID
    def get_deadline_by_id(self, deadline_id):

        for deadline in self.deadlines:

            if deadline["deadline_id"] == deadline_id:
                return deadline

        return None


print("Updated Procedure Engine created successfully.")
```

```python
test_questions = [
    "How can I report misconduct committed by another student?",
    "Where should I submit a written complaint?",
    "Who handles minor misconduct cases?",
    "What happens if the SSO cannot resolve my complaint?",
    "Do I have the right to respond to an allegation?",
    "How long does the CDC have to make a decision?",
    "How can I appeal a disciplinary decision?",
    "What are the valid grounds for an appeal?",
    "How long do I have to submit an appeal?",
    "What can the University Disciplinary Committee do?",
    "How long does the UDC have to communicate its decision?",
    "Who handles academic dishonesty cases?"
]

for question in test_questions:

    print("\n" + "=" * 80)
    print("QUESTION:", question)

    results = procedure_engine.search(question, top_k=3)

    if not results:
        print("No matching procedures found.")
        continue

    for result in results:

        print("\nProcedure:", result["procedure_name"])
        print("ID:", result["procedure_id"])
        print("Section:", result["section_number"])
        print("Page:", result["source_page"])
        print("Score:", result["retrieval_score"])
        print("Matched terms:", result["matched_terms"])
        print("Description:", result["description"])
```
